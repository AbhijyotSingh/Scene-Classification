<b>Scene Classification</b>

<i>In this project, I aim to classify images of natural scenes using CNN</i>

1.<u>Data preprocessing</u>

<i>Importing modules</i>

```python
import tensorflow as tf
```

<i>Preprocessing the training set</i>

```python
training_set=tf.keras.utils.image_dataset_from_directory(
    r"C:\Users\Abhijyot Singh Roda\Desktop\Coding Stuff\Python\Projects\Not so random projects\ML Projects\Datasets\Streets & Shit\seg_train\seg_train",
    label_mode="categorical",
    batch_size=32,
    image_size=(150,150)
)
```

<i>Preprocessing the test set</i>

```python
test_set=tf.keras.utils.image_dataset_from_directory(
    r"C:\Users\Abhijyot Singh Roda\Desktop\Coding Stuff\Python\Projects\Not so random projects\ML Projects\Datasets\Streets & Shit\seg_test\seg_test",
    label_mode="categorical",
    batch_size=32,
    image_size=(150,150)
)
```

2.<u>Building CNN</u>

<i>Initialising CNN</i>

```python
cnn=tf.keras.models.Sequential()
```

<i>Step 1 : Adding the first convolutional layer</i>

```python
cnn.add(tf.keras.layers.Conv2D(kernel_size=3,filters=32,activation="relu"))
```

<i>Step 2 : Adding the first pooling layer</i>

```python
cnn.add(tf.keras.layers.MaxPooling2D(strides=2,pool_size=2))
```

<i>Adding multiple convolutional and pooling layers</i>

```python
for i in range(4):
    cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3,activation="relu"))
    cnn.add(tf.keras.layers.MaxPooling2D(strides=2, pool_size=2))
```

<i>Step 3 : Flattening</i>

```python
cnn.add(tf.keras.layers.Flatten())
```

<i>Step 4 : Full connected</i>

```python
for i in range(5):
    cnn.add(tf.keras.layers.Dense(units=128, activation="relu"))
```

<i>Step 5 : Output layer</i>

```python
cnn.add(tf.keras.layers.Dense(units=6, activation="softmax"))
```

3.<u>Compiling and training the CNN model</u>

<i>Compiling the CNN model</i>

```python
cnn.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
```

<i>Training the CNN model</i>

```python
cnn.fit(x=training_set, validation_data=test_set, epochs=20)
```

4.<u>Model deployment</u> 

```python
cnn.save("CNN.keras")
```
